```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# 1. Feature CSV 로드
df = pd.read_csv("../../kafka_streams/feature_topic_samples/feature_tooldelay_s02.csv", parse_dates=["timestamp"])
df.head()
```

```python
# 규칙 기반 이상 감지
df['rule_alert'] = 0
# Reject streak 2 이상 → 경고
df.loc[df['reject_streak'] >= 2, 'rule_alert'] = 1
# 사이클 지연 > 10% → 경고
df.loc[df['cycle_time_avg'] > 150*1.1, 'rule_alert'] = 1
```

```python
# 통계 기반 이상 감지
features = ['production_interval_avg','reject_count','alarm_count']
window = 5

for col in features:
    df[f'{col}_mean'] = df[col].shift(1).rolling(window).mean()
    df[f'{col}_std'] = df[col].shift(1).rolling(window).std().replace(0, 1e-6)
    df[f'{col}_zscore'] = np.abs(df[col] - df[f'{col}_mean']) / df[f'{col}_std']
    df[f'{col}_zscore'] = df[f'{col}_zscore'].fillna(0)

# threshold = 2
df['stat_alert'] = (df[[f'{col}_zscore' for col in features]] > 2).any(axis=1).astype(int)


# 통합 이상 알람
df['combined_alert'] = df[['rule_alert','stat_alert']].max(axis=1)


# AI Detection Result Visualization (Trigger-based)
plot_df = df.copy()
plot_df["timestamp"] = pd.to_datetime(plot_df["timestamp"])

# Alert columns
alert_cols = ["rule_alert", "stat_alert", "combined_alert"]
alert_labels = {
    "rule_alert": "Rule-based",
    "stat_alert": "Statistical",
    "combined_alert": "Hybrid Alert"
}

# Y positions
y_positions = {
    "rule_alert": 2,
    "stat_alert": 1,
    "combined_alert": 0
}

plt.figure(figsize=(12, 3.8))

for col in alert_cols:
    y = y_positions[col]

    # baseline line
    plt.hlines(
        y=y,
        xmin=plot_df["timestamp"].min(),
        xmax=plot_df["timestamp"].max(),
        linewidth=1,
        alpha=0.3
    )

    # 🔥 핵심: "처음 발생 순간"만 필터링
    trigger_points = plot_df[
        (plot_df[col] == 1) & (plot_df[col].shift(1).fillna(0) == 0)
    ]
    
    if col == "stat_alert":
       trigger_points = trigger_points.head(1)

    # alert points
    plt.scatter(
        trigger_points["timestamp"],
        np.full(len(trigger_points), y),
        s=140,
        marker="o",
        label=alert_labels[col]
    )

# Hybrid 기준 첫 감지 강조
first_hybrid = plot_df[
    (plot_df["combined_alert"] == 1) & 
    (plot_df["combined_alert"].shift(1).fillna(0) == 0)
]

if len(first_hybrid) > 0:
    first_time = first_hybrid["timestamp"].iloc[0]

    plt.axvline(first_time, linestyle="--", linewidth=1.5)

    plt.text(
        first_time,
        1.0,
        "First Detection",
        rotation=90,
        va="top",
        ha="right",
        fontsize=10
    )

# Axis
plt.yticks(
    [2, 1, 0],
    ["Rule-based", "Statistical", "Hybrid"]
)

plt.title("AI Detection Result: Trigger-based Hybrid Alert", fontsize=13, pad=10)
plt.xlabel("Time")
plt.ylabel("Alert Type")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
plt.grid(axis="x", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
# 시각화
plt.figure(figsize=(12,5))
sns.lineplot(data=df, x='timestamp', y='reject_count', marker='o', label='Reject Count')
sns.scatterplot(data=df, x='timestamp', y='reject_count', hue='combined_alert',
                palette={0:'green',1:'red'}, s=120)

plt.title("Event-based Production Flow Alert")
plt.xlabel("Timestamp")
plt.ylabel("Reject Count")
plt.legend()
plt.grid(True)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
plt.xticks(rotation=45)
plt.show()
```

```python
# 히트맵: Rule vs Statistical Alert 비교
alert_map = df[['rule_alert','stat_alert','combined_alert']].T
plt.figure(figsize=(8,2))
sns.heatmap(alert_map, annot=True, cmap='Reds', cbar=False)
plt.title("Rule vs Statistical vs Combined Alert Heatmap")
plt.xlabel("Window (last 5 products)") 
plt.ylabel("Alert Type")
plt.show()
```
